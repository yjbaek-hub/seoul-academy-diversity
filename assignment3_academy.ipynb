# 학원가는 도시의 다양성이 낮을까?

## 요약

- 지표: 학원 입지계수(LQ), 학생 1인당 학원 매출(학원 밀도), 업종 다양성(Gini–Simpson), 용도지역 비율

## 3. 지표 정의, 계산식, 해석

모든 지표의 **분석 단위는 행정동**이고 **공간 단위도 행정동**이다. 기호는 다음과 같다. 동 $i$, 업종 $j$, 동 $i$의 업종 $j$ 매출 $x_{ij}$, 동 전체 매출 $x_i=\sum_j x_{ij}$, 업종 비중 $p_{ij}=x_{ij}/x_i$, 업종 수 $K=63$이다.

### 지표 1. 학원 입지계수 LQ: 학원은 어디에 몰려 있는가
$$LQ_i=\dfrac{x_{i,\text{학원}}/x_i}{x_{\text{학원}}/x}$$
- **변수:** $x_{i,\text{학원}}$는 동 $i$의 일반교습학원 매출이고, $x_{\text{학원}}/x$는 서울 전체 매출 중 학원 비중(3.58%)이다.
- **분석·공간 단위:** 행정동 400개, 2026년 2분기.
- **해석:** 1이면 서울 평균과 같다. **값이 높을수록** 동네 매출에서 학원이 차지하는 비중이 서울 평균보다 큰, 학원에 특화된 동네다(LQ 10이면 서울 평균의 10배). **값이 낮을수록**(1 미만) 학원이 평균보다 적고, 0에 가까우면 학원이 거의 없다.
- **"학원 특화 동"(직접 설계):** LQ가 1.5 이상이면서 학원 비중이 서울 평균보다 10%p 이상 높은 동이다. 원래 작은 업종은 조금만 있어도 LQ가 커 보이므로, 비중 차이 조건을 더해 이를 거른다. 기준값은 분석자가 정했다.
- **자료와 전처리:** 상권분석서비스 추정매출(2026년 2분기). 업종이 20개 이상인 동만 사용한다.

### 지표 2. 학생 1인당 학원 매출(학원 밀도, Density): 학생 수에 비해 학원 매출은 어디에서 가장 큰가 (직접 설계)
$$D_i=\dfrac{x_{i,\text{학원}}}{P_i^{0\text{~}19}}\ \ (\text{만 원/인/분기})$$
- **변수:** $x_{i,\text{학원}}$는 동 $i$의 일반교습학원 매출이고, $P_i^{0\text{~}19}$는 동 $i$의 0~19세 생활인구(평일 평균)로, 학생 연령 인구의 대용치다.
- **논리:** 학원 매출은 동네가 크거나 상권이 클수록 커지므로 그대로 비교하기 어렵다. 동네의 0~19세 생활인구로 나누어 학생 1인당 학원 매출로 바꾸면, 동네 크기의 영향을 줄여 학생 수에 비한 학원 매출을 비교할 수 있다. 거주인구 대신 **생활인구**를 쓴 이유는, 학원은 외부에서 오는 학생을 상대하므로 그 시간에 동네에 있는 학생을 세는 편이 맞기 때문이다.
- **분석·공간 단위:** 행정동 391개(생활인구 500명 이상).
- **해석:** **값이 높을수록** 학생 수에 비해 학원 매출이 큰 동네다. 학원이 그 동네 학생뿐 아니라 외부 학생까지 끌어모으는 집중지로 볼 수 있다. **값이 낮을수록** 학생은 많은데 학원은 적은 동네다. 서울 전체는 약 60만 원, 동 중앙값은 22만 원이다.
- **자료와 전처리:** 매출(2026년 2분기)과 생활인구(2026년 4~6월 평일 평균).

### 지표 3. 업종 다양성 Gini–Simpson: 학원가는 업종이 단조로운가
$$D^{GS}_i=1-\sum_{j=1}^{K}p_{ij}^2,\qquad 0\le D^{GS}\le 1-\tfrac1K$$
- **변수:** $p_{ij}$는 동 $i$ 매출 중 업종 $j$의 비중이다. 무작위로 두 번 골랐을 때 서로 다른 업종일 확률로 읽는다.
- **분석·공간 단위:** 행정동 400개.
- **해석:** **값이 높을수록** 여러 업종이 고르게 섞인 동네이고, **값이 낮을수록** 한두 업종에 쏠린 동네다. 최댓값은 1이 아니라 $1-1/K$(=0.984)이며, 서울 동 중앙값은 0.889다.
- **"학원 제외 다양성"(직접 설계):** 같은 식을 일반교습학원을 제외한 62개 업종으로 다시 계산한다. 학원 비중이 큰 동은 정의상 $\sum p^2$이 커져 다양성이 낮아지는 기계적 효과가 있으므로, 학원을 포함한 값만으로는 상권 전체가 단조로운지 판단할 수 없다. 학원을 제외한 값을 함께 비교해 나머지 상권의 다양성을 따로 확인한다.
- **자료와 전처리:** 지표 1과 같은 매출 자료를 쓴다. 강건성 확인으로 Entropy(수업 지표)도 함께 계산했으며, 두 지표는 거의 같은 순위를 만든다.

### 보조 변수. 용도지역 비율: 도시 구조와의 연결
동마다 시가지 면적 대비 용도지역의 면적 비율을 계산한다. 새 지표가 아니라 위 지표를 해석하기 위한 관련 변수다. 용도지역은 서로 겹치지 않는 세 구분으로 나누며, 나머지는 공업지역(자료상 준공업지역)이다.

| 구분 | 포함하는 용도지역 | 법령상 성격 |
|---|---|---|
| 전용·제1·2종일반주거 | 제1·2종전용주거지역, 제1·2종일반주거지역 | 전용주거는 양호한 주거환경을 보호하는 지역(제1종은 단독주택, 제2종은 공동주택 중심). 제1종일반은 저층주택, 제2종일반은 중층주택 중심 |
| 제3종일반주거 | 제3종일반주거지역 | 중·고층주택 중심 |
| 준주거·상업 | 준주거지역, 중심·일반·근린·유통상업지역 | 준주거는 주거 기능 위주에 일부 상업·업무 기능을 보완하는 지역, 상업은 상업·업무 기능 중심 |

- **분류의 주의:** 준주거지역은 법령상 주거지역의 한 종류지만, 상업·업무 기능이 섞이므로 상업지역과 묶었다. 이 분류는 분석자가 정했다.
- **해석:** 학원 지표와 어떤 용도지역의 상관이 높은지로 학원가가 놓인 토지이용을 읽는다.

```python
def gini_simpson(M):
    """Gini-Simpson 지수 D = 1 - sum(p_k^2): 무작위로 고른 두 결제가 서로 다른 업종일 확률"""
    P = M.div(M.sum(axis=1), axis=0)
    return 1 - (P ** 2).sum(axis=1)

def entropy(M):
    P = M.div(M.sum(axis=1), axis=0).values
    return pd.Series(-np.nansum(np.where(P > 0, P * np.log(P), 0), axis=1) / np.log(N_IND), index=M.index)

ind = pd.DataFrame(index=A.index)
ind["academy_amt"] = A[ACADEMY]
ind["academy_share"] = A[ACADEMY] / A.sum(axis=1)
seoul_share = A[ACADEMY].sum() / A.values.sum()
ind["lq"] = ind["academy_share"] / seoul_share                               # 지표 1. 입지계수
ind["pop_0_19"] = pop["pop_0_19"].reindex(ind.index)
ind["density"] = np.where(ind["pop_0_19"] >= 500, ind["academy_amt"] / ind["pop_0_19"] / 1e4, np.nan)   # 지표 2. 학원 밀도 (만원/인/분기)
ind["gs_all"] = gini_simpson(A)                                              # 지표 3. 다양성 (전체 업종)
ind["gs_noac"] = gini_simpson(A.drop(columns=ACADEMY))                       #         (학원 제외)
ind["ent_all"] = entropy(A)                                                  # 강건성 확인용
ind["ent_noac"] = entropy(A.drop(columns=ACADEMY))
ind["ac_town"] = (ind["lq"] >= 1.5) & (ind["academy_share"] - seoul_share >= 0.10)   # 학원 특화 동

print(f"서울 평균 학원 비중 {seoul_share:.2%} | LQ>1인 동 {int((ind['lq'] > 1).sum())}곳 | LQ≥3인 동 {int((ind['lq'] >= 3).sum())}곳 | 학원 특화 동 {int(ind['ac_town'].sum())}곳")
print(ind[["academy_share", "lq", "density", "gs_all", "gs_noac"]].describe().round(3).to_string())
```

### 4-3. 학원가는 업종이 단조로운가: 공간 패턴

두 지도는 색 구간을 같게 맞췄다. 왼쪽은 63개 업종 전체, 오른쪽은 일반교습학원을 제외한 업종으로 계산한 Gini–Simpson 지수이고, 검은 윤곽선이 학원 특화 동이다.

- 학원 특화 동이 진한 칸(단조로움)에 몰려 있지 않고 진한 칸과 옅은 칸에 섞여 있다. 학원 특화 동이 곧 단조로운 동이라는 공간 패턴은 뚜렷하지 않다.
- 서울에서 업종이 가장 단조로운 동은 학원가가 아니라 노량진2동(0.17), 제기동(0.34), 노량진1동(0.41), 잠실6동(0.45), 독산1동(0.48)이다. 한 업종이 상권 대부분을 차지하는 시장·전문상가 일대다.

```python
# 그림 3 (공간) 업종 다양성(Gini–Simpson): 전체 업종 vs 학원 제외 (같은 색 구간)
vals = pd.concat([gdf["gs_all"], gdf["gs_noac"]]).dropna()
edges = np.quantile(vals, [0, .2, .4, .6, .8, 1.0]); edges[0], edges[-1] = 0, 1
cmap = ["#7f2704", "#d94801", "#fd8d3c", "#fdd0a2", "#fff1e0"]
cls_labels = [f"{edges[i]:.2f} ~ {edges[i + 1]:.2f}" for i in range(5)]
fig, axes = plt.subplots(1, 2, figsize=(18, 9))
for ax, col, title in [(axes[0], "gs_all", "전체 업종 기준"), (axes[1], "gs_noac", "일반교습학원을 제외한 기준")]:
    g = gdf.copy()
    g["cls"] = pd.cut(g[col], bins=edges, labels=False, include_lowest=True)
    g[~g["valid"]].plot(ax=ax, color="#cfcfcf", edgecolor="white", linewidth=0.3)
    for k, c in enumerate(cmap):
        g[g["cls"] == k].plot(ax=ax, color=c, edgecolor="white", linewidth=0.3)
    g[g["ac_town"] == True].boundary.plot(ax=ax, color="black", linewidth=1.1)
    ax.set_title(f"업종 다양성 (Gini–Simpson) — {title}", fontsize=14); ax.set_axis_off()
fig.legend(handles=[Patch(facecolor=c, edgecolor="#999", label=l) for c, l in zip(cmap, cls_labels)]
           + [Patch(facecolor="none", edgecolor="black", linewidth=1.1, label="학원 특화 동 (윤곽선)"), Patch(facecolor="#cfcfcf", label="제외")],
           title="Gini–Simpson (진할수록 단조로움)", loc="lower center", ncol=7, fontsize=11, title_fontsize=12, frameon=False)
fig.suptitle("④ 학원 특화 동은 업종이 단조로운가? 학원을 제외하고 보면?", fontsize=16)
plt.tight_layout(rect=(0, 0.08, 1, 1))
plt.savefig("image/academy3_fig4_gs_maps.png", dpi=150)
plt.show()
```

### 4-5. 학원가는 어떤 토지이용 위에 있는가 (용도지역)

| 용도지역 비율 (시가지 대비, 중앙값) | 학원 특화 동 | 그 외 동 | LQ와의 상관계수 | 학원 밀도와의 상관계수 |
|---|---|---|---|---|
| 전용·제1·2종일반주거 | 42% | 62% | −0.07 | −0.16 |
| **제3종일반주거** | **51%** | **23%** | **+0.39** | **+0.34** |
| 준주거·상업 | 3% | 6% | −0.22 | −0.04 |

- **학원 특화 동은 제3종일반주거의 비율이 2배 이상 높다**(51% 대 23%). LQ와의 상관도 제3종일반주거에서만 뚜렷하고(+0.39), 학원 밀도도 같은 방향이다(+0.34).
- 같은 주거지라도 **전용·제1·2종일반주거는 오히려 학원 특화 동에서 낮고**(42% 대 62%) 상관도 없거나 약하다(−0.07, −0.16). 준주거·상업은 3% 대 6%로 학원 특화 동에서 낮다(상관계수 −0.22).
- 동을 유형으로 나누면, 전용·일반주거 합계가 70% 이상인 338개 동 중 학원 특화 동이 28곳이고, 준주거·상업이 40% 이상인 19개 동에는 **한 곳도 없다.**
- LQ 상위 6개 동의 제3종일반주거 비율은 중계1동 76%, 대치1동 72%, 반포2동 72%, 돈암1동 65%, 목5동 62%, 신정7동 52%다(위 출력). 이 6곳은 준주거·상업 비율이 0~10%다.
- 단, **제3종일반주거가 많다고 학원가가 되지는 않는다.** 전용·일반주거 합계가 70% 이상인 338개 동 중 학원 특화 동은 28곳(약 8%)뿐이다. 또 세 구분의 비율은 합이 100%에 가까워(공업지역 제외) 한 구분이 커지면 다른 구분이 작아진다(전용·일반주거 합계와 준주거·상업 비율의 상관계수 −0.83). 그래서 상관이 가장 뚜렷한 제3종일반주거를 중심으로 해석하는 것이 안전하다.
- 업종 다양성은 어느 용도지역과도 거의 무관하다(상관계수 −0.00~+0.11). 토지이용으로 설명되는 것은 학원의 집중이고, 다양성은 아니다.

```python
# 보조 변수: 용도지역 비율 (도시 구조·토지이용과의 연결용)
z = gpd.read_file(glob.glob("shp/569_UQ111_용도지역(도시지역)_202609/*.shp")[0], encoding="cp949").to_crs(5179)
z = z[z["ATRB_SE"].notna()].copy(); z["geometry"] = z.geometry.make_valid()
def zone_cls(c):
    if c.startswith("UQA11") or c in ("UQA121", "UQA122", "UQA124", "UQA129"): return "전용·제1·2종일반주거"
    if c == "UQA123": return "제3종일반주거"
    if c in ("UQA130", "UQA190") or c.startswith("UQA2"): return "준주거·상업"
    if c.startswith("UQA3"): return "공업"
    return None
z["cls"] = z["ATRB_SE"].map(zone_cls); z = z[z["cls"].notna()]
zd = z.dissolve(by="cls").reset_index()[["cls", "geometry"]]
ov = gpd.overlay(gj[["ADSTRD_CD", "geometry"]], zd, how="intersection", keep_geom_type=True)
ov["area"] = ov.geometry.area
ar = ov.pivot_table(index="ADSTRD_CD", columns="cls", values="area", aggfunc="sum", fill_value=0)
K = ["전용·제1·2종일반주거", "제3종일반주거", "준주거·상업", "공업"]
for c in K:
    if c not in ar.columns: ar[c] = 0
urban = ar[K].sum(axis=1)
Z = (ar[K[:3]].div(urban, axis=0))[urban > 0]   # 서로 겹치지 않는 3개 구분 (나머지는 공업지역)
d = ind2.join(Z, how="inner")
print(f"용도지역과 결합된 동 {len(d)} | 시가지(전용·일반주거+준주거+상업+공업) 면적 대비 비율")
print("서울 동 중앙값(%):", {c: round(d[c].median() * 100) for c in Z.columns})
print("\n[학원 특화 동 vs 그 외] 중앙값(%)")
for c in Z.columns:
    t, r = d[d["ac_town"]][c], d[~d["ac_town"]][c]
    print(f"  {c}: {t.median() * 100:.0f} vs {r.median() * 100:.0f}")
print("\n[상관계수(순위 기준)]        LQ    학원 밀도   지니심슨(전체)  지니심슨(학원 제외)")
for c in Z.columns:
    print(f"  {c:<10} {spearmanr(d.lq, d[c])[0]:+.2f}   {spearmanr(d.dropna(subset=['density']).density, d.dropna(subset=['density'])[c])[0]:+.2f}      {spearmanr(d.gs_all, d[c])[0]:+.2f}          {spearmanr(d.gs_noac, d[c])[0]:+.2f}")
d["주거합계"] = d["전용·제1·2종일반주거"] + d["제3종일반주거"]
d["type"] = np.where(d["주거합계"] >= 0.7, "전용·일반주거 합계 70%↑", np.where(d["준주거·상업"] >= 0.4, "준주거·상업 40%↑", "혼합"))
print("\n[유형별] 동 수 / 학원 특화 동 수 / LQ 중앙값")
print(d.groupby("type").agg(동=("lq", "size"), 학원특화동=("ac_town", "sum"), LQ중앙값=("lq", "median")).round(2).to_string())
print("전용·일반주거 합계와 준주거·상업 비율의 상관(서로 반대로 움직이는 정도):", round(spearmanr(d["주거합계"], d["준주거·상업"])[0], 2))
print("학원 LQ 상위 6동의 전용·제1·2종 / 제3종 / 준주거·상업 (%):", ", ".join(f"{r.dong_nm} {r['전용·제1·2종일반주거'] * 100:.0f}/{r['제3종일반주거'] * 100:.0f}/{r['준주거·상업'] * 100:.0f}" for _, r in d.sort_values("lq", ascending=False).head(6).iterrows()))
print("학원 특화 동의 전용·일반주거 합계 최소 %.0f%%, 중앙값 %.0f%%" % (d[d["ac_town"]]["주거합계"].min() * 100, d[d["ac_town"]]["주거합계"].median() * 100))
print("학원 특화 동의 공업지역 비율 중앙값 %.0f%%, 그 외 동 %.0f%%" % ((1 - d[d["ac_town"]][list(Z.columns)].sum(axis=1)).median() * 100, (1 - d[~d["ac_town"]][list(Z.columns)].sum(axis=1)).median() * 100))
print("지정 면적(㎢): " + ", ".join(f"{k} {ar[k].sum() / 1e6:.0f}" for k in K))
```

```python
# 그림 5 (비공간) 학원 특화 동은 어떤 토지이용에 있는가: 용도지역 비율
fig, axes = plt.subplots(1, 3, figsize=(16, 5.2))
for ax, c in zip(axes, ["전용·제1·2종일반주거", "제3종일반주거", "준주거·상업"]):
    t, r = d[d["ac_town"]][c] * 100, d[~d["ac_town"]][c] * 100
    bp = ax.boxplot([t, r], tick_labels=[f"학원 특화 동\n({len(t)}곳)", f"그 외 동\n({len(r)}곳)"], patch_artist=True, widths=0.55)
    for patch, col in zip(bp["boxes"], ["#f59e6c", "#9ecae1"]):
        patch.set_facecolor(col)
    rho = spearmanr(d["lq"], d[c])[0]
    ax.set_title(f"{c} 비율 (%)\n중앙값 {t.median():.0f} vs {r.median():.0f} · LQ와 상관계수 {rho:+.2f}", fontsize=12)
    ax.grid(alpha=0.25, axis="y")
fig.suptitle("⑥ 학원 특화 동은 제3종일반주거 비율이 높고 준주거·상업 비율은 낮다", fontsize=14)
plt.tight_layout()
plt.savefig("image/academy3_fig6_zoning.png", dpi=150)
plt.show()
```

## 5. 종합 해석

**연구 질문에 대한 답.** 학원가는 도시의 다양성이 낮은가? **전체 업종으로 보면 다소 낮지만, 그것은 학원이 매출의 큰 몫을 차지하기 때문이며 학원을 제외하면 서울의 다른 동과 같다.** 학원가는 단조로운 동네라기보다, 일반적인 주거 근린 상권에 학원이라는 큰 업종이 더해진 동네에 가깝다(이 마지막 표현은 추정이다). 과제가 묻는 네 가지 질문에 차례로 답한다.

**① 값이 높은 곳과 낮은 곳은 어디인가**
- **학원 비중(LQ)이 가장 높은 곳**은 목5동, 중계1동, 대치1동, 돈암1동, 반포2동이다. **학원 밀도(학생 1인당)가 가장 높은 곳**은 대치4동, 명일2동, 대치2동, 대치1동이다(노량진2동은 예외). 같은 학원가라도 학원 비중이 가장 큰 동네와 학생 1인당 학원 매출이 가장 큰 동네는 일부 다르다.
- **업종 다양성이 가장 낮은 곳**은 학원가가 아니라 노량진·제기동·독산1동 같은 시장·전문상가 일대다.

**② 지표 간에 비슷하거나 다른 공간 패턴이 나타나는가**
- **비슷한 패턴:** LQ와 학원 밀도는 같은 동네를 가리킨다(상관계수 +0.83). 둘 다 강남·서초·송파 띠와 양천, 노원, 성북의 거점에 몰린다.
- **다른 패턴:** 업종 다양성(Gini–Simpson)은 학원 지표와 같은 군집을 만들지 않고, 학원 특화 동과 그 외 동의 차이도 작다(학원을 제외하면 차이가 거의 없음). 학원이 많다는 것과 상권이 단조롭다는 것은 별개의 현상이다.
- **갈리는 동:** 중계1동(LQ 높음, 밀도 낮음), 노량진2동(밀도 높음, LQ 낮음)처럼 두 지표가 다르게 읽히는 동이 있어, 두 지표를 함께 보는 의미가 있다.

**③ 관찰된 패턴을 도시 구조나 토지이용 특성과 어떻게 연결할 수 있는가**
- 학원 특화 동 29곳의 전용·일반주거 합계 비율은 최소 68%, 중앙값 97%이고, 준주거·상업 비율(중앙값 3%)은 서울 다른 동(6%)의 절반이다. 학원 LQ는 **제3종일반주거 비율과 가장 강한 양의 상관(+0.39)**을, 준주거·상업 비율과는 음의 상관(−0.22)을 보이고, 전용·제1·2종일반주거와는 거의 무관하다(−0.07).
- 제3종일반주거는 법적으로 중·고층 주택이 중심이 되는 지역이므로, 학원가는 준주거·상업지구가 아니라 **중·고층 주택 중심의 주거지**에 모인다고 해석할 수 있다. 같은 주거지라도 저층·중층 위주의 전용·제1·2종일반주거에서는 그렇지 않았다. 대치·반포(강남권), 목동·신정(양천), 중계(노원), 돈암(성북)이 그 예다.
- 학원 특화 동 안에서도 유형이 갈린다. 목5동·중계1동은 학원을 제외하면 평범한 상권이어서 주거지형 학원가이고, 돈암1동은 학원 외 상권도 약하다.

**④ 분석 결과만으로는 설명할 수 없는 점은 무엇인가**
1. **추정 매출이다.** 카드 결제를 바탕으로 한 추정값이라 계좌이체가 많은 학원은 실제보다 작게 잡혔을 수 있다. 매출은 분기별 합계로 제공되며, 학원 밀도는 이 분기 매출을 기준으로 한 값이다. 동 사이 비중(LQ, 다양성)은 비율이라 매출의 기간 단위와 무관하다.
2. **점포 소재지 기준이다.** 매출은 학원이 있는 동에 잡히고 학생이 사는 동과는 관계가 없다. 학원 밀도의 분모(생활인구)는 하루 8개 시간대의 평균이라 학교나 집에 있는 학생도 포함하며, 학원에 다니는 학생만 센 것이 아니다.
3. **"학원" 업종 정의의 한계가 있다.** 일반교습학원이 청소년 입시·보습 학원을 대표하지만, 영어 어학원은 외국어학원으로 분류되어 빠졌을 수 있고 성인 대상 학원이 섞여 있을 수 있다(노량진2동이 그 예일 수 있다). 업종 분류의 세부 정의는 확인하지 못했다.
4. **"학원 특화 동"의 정의가 결과에 영향을 준다.** 기준(LQ 1.5, +10%p)은 분석자가 정했고, 정의에 따라 일부 비교 결과가 달라질 수 있다(예: 다양성 비교).
5. **행정동 경계의 영향(MAUP)이 있다.** 대치동은 대치1·2·4동으로 나뉘어 있어서 하나로 묶으면 값이 달라질 수 있다.
6. **용도지역은 법적 구분이다.** 주거지역 안에도 상가가 있고 학원은 상가 건물에 입지하므로, 용도지역과의 상관은 주거지에 둘러싸인 상가를 뜻할 수 있다. 이 해석은 자료로 검증하지 않았다. 세 구분의 비율이 서로 반대로 움직이는 점도 고려해야 한다. 준주거지역을 상업지역과 묶은 것은 분석자의 분류이며, 다르게 분류했을 때의 결과 차이는 확인하지 않았다.
7. **한 분기(2026년 2분기)의 단면이다.** 학원가가 어떻게 생기고 변해 왔는지, 아파트가 학원을 부르는지 학원이 아파트 가치를 높이는지는 알 수 없다. 상관관계일 뿐 인과관계가 아니다.
8. **생활밀착 63개 업종만 다뤘다.** 백화점, 사무실, 온라인 소비는 없고, 다양성은 업종 구성으로만 쟀다. 거리의 분위기나 소음 같은 체감은 반영하지 못했다.
